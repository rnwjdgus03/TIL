```python
from huggingface_hub import InferenceClient

client = InferenceClient(
    provider="featherless-ai",
    api_key=""   # 여기에 새로 발급받은 HF 토큰 입력
)

prompt = """
You are a careful reasoning assistant.

Use the following Structured Chain of Thought.

Step 1. Understand the problem.
Step 2. Identify the given information.
Step 3. Make a plan.
Step 4. Solve the problem step by step.
Step 5. Verify your reasoning.
Step 6. Give the final answer.

Problem:
철수가 책 11권을 샀고, 한 권의 가격은 18,000원입니다. 행사는 3권당 10% 할인입니다. 6권까지는 3권씩 묶음 할인이 적용되고 나머지 한 권에는 적용되지 않습니다. 또한, 결제 금액의 5%를 포인트로 적립받았습니다. 철수가 지불한 금액과 적립된 포인트 금액은 얼마인가요?
"""

response = client.chat.completions.create(
    model="meta-llama/Llama-3.1-8B-Instruct",
    messages=[
        {"role": "user", "content": prompt}
    ],
    temperature=0.2,
    max_tokens=1024,
)

print(response.choices[0].message.content)
```

### Zero-Shot Chain-of-Thought (CoT) Prompt

이번에는 모델에게 구체적인 단계를 지시하지 않고, '단계별로 생각해보자'는 지시만으로 추론 과정을 유도하는 Zero-Shot CoT 방식을 사용합니다.

```python
zero_shot_prompt = """
Problem:
철수가 책 11권을 샀고, 한 권의 가격은 18,000원입니다. 행사는 3권당 10% 할인입니다. 6권까지는 3권씩 묶음 할인이 적용되고 나머지 한 권에는 적용되지 않습니다. 또한, 결제 금액의 5%를 포인트로 적립받았습니다. 철수가 지불한 금액과 적립된 포인트 금액은 얼마인가요?

Let's think step by step.
"""

zero_shot_response = client.chat.completions.create(
    model="meta-llama/Llama-3.1-8B-Instruct",
    messages=[
        {"role": "user", "content": zero_shot_prompt}
    ],
    temperature=0.2,
    max_tokens=1024,
)

print(zero_shot_response.choices[0].message.content)
```
